## 1. GENERAR UN DATAFRAME CON LOS DATOS DEL FICHERO

```python
import pandas as pd
titanic = pd.read_csv('/kaggle/input/datasets/rodrigogarciah/titanic/titanic.csv')

print(titanic)
```

## 2. MOSTRAR POR PANTALLA LAS DIMENSIONES DEL DATAFRAME, EL NÚMERO DE DATOS QUE CONTIENE, LOS NOMBRES DE SUS COLUMNAS Y FILAS, LOS TIPOS DE DATOS DE LAS COLUMNAS, LAS 10 PRIMERAS FILAS Y LAS 10 ÚLTIMAS FILAS

```python
#mostrar por pantalla las dimensiones del DataFrame 
print('Dimensiones', titanic.shape)
print('Numero de elementos', titanic.size)
print('Nombres de columnas', titanic.columns)
print('Nombres de filas', titanic.index)
print('Tipos de datos', titanic.dtypes)
print('Primeras 10 filas', titanic.head(10))
print('Ultimas 10 filas', titanic.tail(10))
```

## 4. MOSTRAR POR PANTALLA LAS FILAS PARES DEL DATAFRAME

## 5. MOSTRAR POR PANTALLA LOS NOMBRES DE LAS PERSONAS QUE IBAN EN PRIMERA CLASE ORDENADAS ALFABETIFICAMENTE

```python
PrimeraClase = titanic[titanic["Pclass"]==1]['Name'].sort_values()
print(PrimeraClase)
```

## 6. MOSTRAR POR PANTALLA EL PORCENTAJE DE LAS PERSONAS QUE SOBREVIVIERON Y MURIERON 

(CUANTOS HOMBRES Y MUJERES)

```python
print(titanic['Survived'].value_counts()/titanic['Survived'].count() *100)
conteo_sexo = titanic["Sex"].value_counts()
print(conteo_sexo)
```

## 7. MOSTRAR POR PANTALLA EL PORCENTAJE DE PERSONAS QUE SOBREVIVIERON EN CADA CLASE

```python
supervivencia_clase = titanic.groupby("Pclass")["Survived"].mean()*100
print(supervivencia_clase)
```

## 8. ELIMINAR DEL DATAFRAME LOS PASAJEROS CON EDAD DESCONOCIDA

```python
titanic =titanic.dropna(subset = ["Age"])
print(len(titanic))
```

## 9. MOSTRAR POR PANTALLA LA EDAD MEDIA DE LAS MUJERES QUE VIAJABAN EN CADA CLASE

```python
edad_media_mujeres = titanic[titanic["Sex"] == "female"].groupby("Pclass")["Age"].mean()

print(edad_media_mujeres)
```

## 10. AÑADIR UNA NUEVA COLUMNA BOOLEANA PARA VER SI EL PASAJERO ERA MENOR DE EDAD O NO

```python
titanic["Es_Menor"]= titanic["Age"] < 18
print(titanic[["Name", "Age", "Es_Menor"]].head(10))
```

## 11. MOSTRAR POR PANTALLA EL PORCENTAJE DE MENORES Y MAYORES DE EDAD QUE SOBREVIVIERON EN CADA CLASE

```python
titanic_edad = titanic.dropna(subset=["Age"]).copy()

titanic_edad["Grupo_Edad"] = "Mayor"
titanic_edad.loc[titanic_edad["Age"] < 18, "Grupo_Edad"] = "Menor"

porcentaje = (
    titanic_edad.groupby(["Pclass", "Grupo_Edad"])["Survived"].mean() * 100
)

print(porcentaje)
```

## 13. IMPUTADO 

```python
titanic_imputado = titanic.copy()

# Calcular estadísticas antes de imputar
media = titanic['Age'].mean()
mediana = titanic['Age'].median()
moda = titanic['Age'].mode()[0]

print("Estadísticas de age (sin nulos):")
print(f"Media  :{media:.2f} años")
print(f"Mediana:{mediana:.1f} años")
print(f"Moda   :{moda:.1f} años")

# Rellenar con la mediana 
titanic_imputado['Age'] = titanic_imputado['Age'].fillna(mediana)

# Convertir a entero
titanic_imputado['Age'] = titanic_imputado['Age'].astype(int)

print(f"\nDespués de imputar con mediana ({mediana}):")
print(f"Tipo   :{titanic_imputado['Age'].dtype}")
print(f"Nulos  :{titanic_imputado['Age'].isnull().sum()}")
print(f"Muestra:{titanic_imputado['Age'].head(10).tolist()}")


edad_media_mujeres = titanic[titanic["Sex"] == "female"].groupby("Pclass")["Age"].mean()

print(edad_media_mujeres)

print(titanic_imputado)
```

## 14. TITANICRODRI

```python
titanicRodri = titanic.copy()

# Nombres de columnas Index(['PassengerId', 'Survived', 'Pclass', 'Name', 'Sex', 'Age', 'SibSp',
   #    'Parch', 'Ticket', 'Fare', 'Cabin', 'Embarked'],

titanicRodri = titanicRodri.rename(
    columns={
        "PassengerId": "IDPasajero",
        "Survived": "¿Sobrevivio?",  
        "Pclass": "Clase", 
        "Name": "Nombre",
        "Sex": "Genero",
        "Age": "Edad",
        "SibSp": "Hermanos/Conyuge",
        "Parch": "Padres/Hijos",
        "Ticket": "Boleto",
        "Fare": "Tarifa",
        "Cabin": "Habitación",
        "Embarked": "Puerto de embarcación"
    }
)

print("Nombres de columnas:", titanicRodri.columns)
```
